# BoxPlot

```python
f = Pivot_table_1['F'].tolist()
m = Pivot_table_1['M'].tolist()

plt.figure()
plt.boxplot([m, f], tick_labels=['Male','Female'], vert=False)
plt.xlabel('Orders')
plt.title('Horizontal Box Plot - Gender')
plt.show()
```

# ScatterPlot

```python
plt.scatter(df['Age'], df['Amount'], c=df['Amount'], cmap='plasma', alpha=0.5)
plt.colorbar(label='Amount')
plt.xlabel('Age')
plt.ylabel('Amount')
plt.title('Age vs Amount - Color Gradient')
plt.show()
```
